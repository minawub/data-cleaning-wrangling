```python
# --- Exercise 18: Missing Values Detection ---
missing_df = pd.DataFrame({
    "A": [1, 2, np.nan, 4, 5],
    "B": [np.nan, "X", "Y", None, "Z"],
    "C": [10.5, np.nan, 30.2, np.nan, 50.1]
})
print("=== Exercise 18 ===")
print("Missing per column:\n", missing_df.isnull().sum())
print("Total Missing:", missing_df.isnull().sum().sum(), "\n")

# --- Exercise 19: Handling Missing Values (Mean vs Median) ---
num_data = pd.DataFrame({"val": [10, 20, np.nan, 40, 100, np.nan, 30]})
print("=== Exercise 19 ===")
print("Filled with Mean:", num_data["val"].fillna(num_data["val"].mean()).values)
print("Filled with Median:", num_data["val"].fillna(num_data["val"].median()).values, "\n")

# --- Exercise 20: Duplicate Records ---
df_dups = pd.DataFrame({
    "id": [1, 2, 2, 3, 4, 1],
    "item": ["Apple", "Banana", "Banana", "Orange", "Grape", "Apple"]
})
print("=== Exercise 20 ===")
print("Duplicates Removed:")
display(df_dups.drop_duplicates())

# --- Exercise 21: Standardizing Text Data ---
cities = ["addis ababa", "ADDIS ABABA", "Addis Ababa", "adDis abaBa"]
clean_cities = [c.strip().title() for c in cities]
print("\n=== Exercise 21 ===")
print("Standardized Cities:", clean_cities, "\n")

# --- Exercise 22: Correcting Data Entry Errors ---
ages = [21, 25, 230, 30, 19, -5, 27]
corrected_ages = [age if (0 <= age <= 120) else np.nan for age in ages]
print("=== Exercise 22 ===")
print("Corrected Ages (Invalid as NaN):", corrected_ages)
```

```python
# --- Exercise 23: Cleaning Customer Dataset ---
raw_customers = pd.DataFrame({
    "customer_id": [101, 102, 102, 103, 104],
    "name": [" abel ", "Sara", "Sara", None, "JOHN "],
    "city": ["addis ababa", "ADDIS ABABA", "ADDIS ABABA", "Hawassa", "adDis abaBa"],
    "spend": [500, np.nan, np.nan, 750, 1200]
})

clean_cust = raw_customers.drop_duplicates().copy()
clean_cust["name"] = clean_cust["name"].fillna("Unknown").str.strip().str.title()
clean_cust["city"] = clean_cust["city"].str.strip().str.title()
clean_cust["spend"] = clean_cust["spend"].fillna(clean_cust["spend"].median())

print("=== Exercise 23: Cleaned Customers ===")
display(clean_cust)

# --- Exercise 24: Outlier Detection (IQR) ---
sales = pd.DataFrame({"sales_amount": [120, 130, 125, 135, 128, 140, 1050, 115, 132, -200]})
Q1 = sales["sales_amount"].quantile(0.25)
Q3 = sales["sales_amount"].quantile(0.75)
IQR = Q3 - Q1
lower, upper = Q1 - 1.5 * IQR, Q3 + 1.5 * IQR

outliers = sales[(sales["sales_amount"] < lower) | (sales["sales_amount"] > upper)]
print("\n=== Exercise 24: Outliers ===")
print("Detected Outliers:\n", outliers["sales_amount"].values)

# --- Exercise 25: Full Pipeline ---
unclean_df = pd.DataFrame({
    "id": [1, 2, 2, 3, 4, 5, 6],
    "name": [" abel", "sara ", "sara ", None, "JOHN", "Helen", "marta"],
    "department": ["cs", "IT", "IT", "se", "CS", "it", None],
    "salary": [4000, 4500, 4500, np.nan, 5000, 4800, 85000],
    "age": [25, 230, 230, 28, 30, -4, 27]
})

def full_pipeline(df):
    d = df.drop_duplicates().copy()
    d["age"] = d["age"].apply(lambda x: x if (0 <= x <= 120) else np.nan)
    d["name"] = d["name"].fillna("Unknown").str.strip().str.title()
    d["department"] = d["department"].fillna("Unknown").str.strip().str.upper()
    d["age"] = d["age"].fillna(d["age"].median())
    d["salary"] = d["salary"].fillna(d["salary"].median())
    
    q1, q3 = d["salary"].quantile(0.25), d["salary"].quantile(0.75)
    iqr = q3 - q1
    l_bound, u_bound = q1 - 1.5 * iqr, q3 + 1.5 * iqr
    
    d["salary"] = np.where(d["salary"] > u_bound, u_bound, d["salary"])
    d["salary"] = np.where(d["salary"] < l_bound, l_bound, d["salary"])
    return d

print("\n=== Exercise 25: Final Cleaned Dataset ===")
display(full_pipeline(unclean_df))
```
